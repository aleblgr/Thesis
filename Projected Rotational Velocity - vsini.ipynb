# Projected rotational velocities (vsini)

Quelccaya is a useful tool for determining projected rotational velocities throughout the spectral synthesis method.
Quelccaya requires q2 to create a model atmosphere. To install it, simply follow the instructions at https://github.com/astroChasqui/q2

```python
from Quelccaya import quel
import numpy as np
#%matplotlib notebook
%matplotlib inline
```

### Structure Files

Quelccaya requires a file named input_data. Within this file, you will find the following archives:

1) abds.csv ---> a file to change the abundance of iron and nitrogen to measure vsini
2) abds_Li.csv ---> a file mainly to change the abundance of Li and other elements
3) lisun7.moog ---> a line list for the synthesis
4) syn.lin ---> a line list for vmacro determination."

### Stellar Parameters

To create a model atmosphere, it is necessary to declare the star's stellar parameters. The 'starid' below is the name of the observed spectrum in FITS format.

```python
starid   = 'K2-111_norm.fits'    #name of the spectra
teff     = 5770
err_teff = 16
logg     = 4.29
err_logg = 0.036
feh      = -0.494
err_feh  = 0.013
vt       = 1.09
err_vt   = 0.03
age      = 11.80
err_age  = 0.44
mass     = 0.9
err_mass = 0.005    
resol    = 140000    #resolution of the instrument
SO = 200             #signal-to-noise ratio
```

### Line 6027.05

```python
line        = 6027.05      #line to be synthetized and listed in syn.li
wl_start    = line - 0.85   #left wavelength limit
wl_end      = line + 0.85  #right wavelength limit
vsini       = 1.1          #value to be calculated

#shifts
y_shift_mult = 0.005    #flux shift applied to the observed spectrum
wl_shift     = -0.008    #wavelength shift applied to the observed spectrum

vsini1, vmac1 = quel.vsini(starid=starid, teff=teff, logg=logg, feh=feh, vt=vt, wl_start=wl_start, wl_end=wl_end, line=line, 
         resol=resol, vsini=vsini, y_shift_mult=y_shift_mult, wl_shift=wl_shift)
```

```python
print('vsini = ', vsini1, ' vmac = ', vmac1)
```
